```python
#6
#6.1 Input data
print(f"Informasi Data")
nama=input('Nama Lengkap\t: ')
usia=int(input('Usia\t\t: '))
kontak=int(input('Kontak\t\t: '))
alamat=input('Alamat\t\t: ')
kampus=input('Nama Kampus\t: ')
fakultas=input('Nama Fakultas\t: ')
departemen=input('Nama Departemen\t: ')
NRP=int(input('NRP\t\t: '))
angkatan=int(input('Angkatan\t: '))
sks=int(input('Jumlah SKS\t: '))
matkul=input('Nama Mata Kuliah: ')
dream_job=input('Target Karier\t: ')

#6.2 Output data
WIDTH = 58
INNER = WIDTH - 2  # lebar isi di antara dua karakter '|'

print(' ')
print(f"Informasi Dashbor")
print('='*WIDTH)
print(f"|{'DASBOR PROFIL MAHASISWA & RENCANA KARIR':^{INNER}}|")
print('='*WIDTH)
print(f"|{'INFORMASI PERSONAL':^{INNER}}|")
print('-'*WIDTH)
print(f"|{' Nama Lengkap  : ' + nama:<{INNER}}|")
print(f"|{' Usia & Kontak : ' f'{usia} Tahun | {kontak}':<{INNER}}|")
print(f"|{' Alamat Asal   : ' + alamat:<{INNER}}|")
print('-'*WIDTH)
print(f"|{'INFORMASI AKADEMIK':^{INNER}}|")
print('-'*WIDTH)
print(f"|{' Kampus        : ' + kampus:<{INNER}}|")
print(f"|{' Fakultas/Dept : ' + f'{fakultas} / {departemen}':<{INNER}}|")
print(f"|{' NRP/Angkatan  : ' + f'{NRP} / {angkatan}':<{INNER}}|")
print(f"|{' SKS & Matkul  : ' + f'{sks} | {matkul}':<{INNER}}|")
print('='*WIDTH)
print(f"|{'TARGET KARIER (DREAM JOB)':^{INNER}}|")
print(f"|{dream_job:^{INNER}}|")
print('='*WIDTH)
```
